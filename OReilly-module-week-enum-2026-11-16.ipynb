# What are enums?

We're used to defining variables in our programs. If I want to set the price of something to 50, I can say

    price = 50

But what if I have a bunch of names and values that I'll want to set? I could define a bunch of variables and values together:

    January = 1
    February = 2
    March = 3

This will work! But there are a few problems with it:

1. The biggest one is that there is no connection between `January` and `February`. Both are just variables that we've defined. There's no semantic connection.
2. How can I compare them with one another? Really, we just have integers here. So we can compare them, but there's nothing "month-ish" about them.
3. In modern Python, especially when we're working on big projects, or projects with many people, or agentic coding, we want to use type hints. And what kinds of type hints could I use here, in this example? I'd have to say that a function that wants to get a month really gets an `int`. I could restrict it (with Pydantic) to say that it must be between 1-12, but that's still not quite the same as saying it has to be a month.
