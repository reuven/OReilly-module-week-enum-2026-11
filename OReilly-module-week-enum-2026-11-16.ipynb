# What are enums?

We're used to defining variables in our programs. If I want to set the price of something to 50, I can say

    price = 50

But what if I have a bunch of names and values that I'll want to set? I could define a bunch of variables and values together:

    January = 1
    February = 2
    March = 3

This will work! But there are a few problems with it:

1. The biggest one is that there is no connection between `January` and `February`. Both are just variables that we've defined. There's no semantic connection.
2. How can I compare them with one another? Really, we just have integers here. So we can compare them, but there's nothing "month-ish" about them.
3. In modern Python, especially when we're working on big projects, or projects with many people, or agentic coding, we want to use type hints. And what kinds of type hints could I use here, in this example? I'd have to say that a function that wants to get a month really gets an `int`. I could restrict it (with Pydantic) to say that it must be between 1-12, but that's still not quite the same as saying it has to be a month.

# For a long time, we used dicts

Everyone in the Python world loves dictionaries. Why not? Dicts are super fast, easy to work with, friendly, etc.

I can create a `MONTHS` dict (using all caps, because it's a "constant"), and all of the names and values:

```python
MONTHS = {'January':1, 'February':2, 'March':3, 'April':4, 'May':5, 'June':6}

```

```python
MONTHS['January']  # this works well
```

```python
favorite_month = input('Enter a month: ').strip()

if favorite_month in MONTHS:
    print(f'You like {favorite_month}, which has a number of {MONTHS[favorite_month]}.')
else:
    print(f'{favorite_month} is made up!')
```

# Enums to the rescue!

Other languages had enums ("enumerated types") for many years; probably about 10 years ago, Python got them, too.

Enums are actually (mostly, or even completely) just based on Python's classes.

```python
from enum import Enum     # "enum" is a module, and "Enum" is a class

type(Enum)
```

# To create an enum

1. I define a new class
2. My class inherits from Enum
3. I set class attributes to be names and values

```python
class Month(Enum):
    JANUARY = 1
    FEBRUARY = 2
    MARCH = 3
    APRIL = 4
    MAY = 5
    JUNE = 6
    JULY = 7
```

When we define an Enum class, we are *not* looking for new ways to store data, or even for custom functionality via methods. (You can define methods, and sometimes that's useful... but usually not.) THe whole point of the enum is to give us a new class whose attributes have a relationship, and can be classified in a single way.

```python
type(Month.JANUARY)
```

```python
Month.JANUARY == 1
```

```python
class OtherThing(Enum):
    X = 1
    Y = 2
    Z = 3
```

```python
Month.JANUARY == OtherThing.X
```
